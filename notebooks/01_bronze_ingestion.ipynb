```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType,
    DoubleType
)
```

```python
raw_path = "/Volumes/workspace/hdb_pyspark/raw_data/hdb_resale_2017_onwards.csv"
```

```python
hdb_schema = StructType([
    StructField("month", StringType(), True),
    StructField("town", StringType(), True),
    StructField("flat_type", StringType(), True),
    StructField("block", StringType(), True),
    StructField("street_name", StringType(), True),
    StructField("storey_range", StringType(), True),
    StructField("floor_area_sqm", DoubleType(), True),
    StructField("flat_model", StringType(), True),
    StructField("lease_commence_date", IntegerType(), True),
    StructField("remaining_lease", StringType(), True),
    StructField("resale_price", DoubleType(), True)
])
```

```python
df_raw = (
    spark.read
    .option("header", True)
    .schema(hdb_schema)
    .csv(raw_path)
)
```

```python
df_raw.printSchema()
```

```python
df_raw.show(5, truncate=False)
```

```python
from pyspark.sql.functions import current_timestamp

df_bronze = df_raw.withColumn(
    "ingested_at",
    current_timestamp()
)

(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.hdb_pyspark.bronze_hdb_resale")
)
```
